# UAP Flight Characteristics Analysis
This notebook analyzes UAP flight data from Pentagon reports (e.g., Tic-Tac, Gimbal).
It calculates acceleration, velocity, and identifies anomalies.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load flight data
flight_data = pd.read_csv('data/raw/uap_flight_data.csv')

# Display the first few rows
flight_data.head()
```

```python
# Plot UAP velocities
plt.figure(figsize=(10, 6))
plt.hist(flight_data['velocity_mach'], bins=10, color='skyblue', edgecolor='black')
plt.title('Distribution of UAP Velocities (Mach)')
plt.xlabel('Velocity (Mach)')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

```python
# Identify anomalies (acceleration > 100 Gs)
anomalies = flight_data[flight_data['acceleration_g'] > 100]
anomalies[['date', 'location', 'acceleration_g', 'anomaly']]
```
